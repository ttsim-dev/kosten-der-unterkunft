# Germany by Gemeinde — prototype map

Choropleth of all ~11,000 German municipalities, filled with **placeholder data**.
Run all cells to view the interactive map inline.

The join keys on a synthetic per-feature `fid`, not on names (which repeat across
Germany). For real data, join on the official AGS in `gem_code`.

```python
from kdu.config import DATA
from kdu.geodata import load_geojson
from kdu.maps import build_choropleth, build_fake_frame

gemeinden = load_geojson(DATA / "gemeinden.geo.json")
len(gemeinden["features"])
```

```python
frame = build_fake_frame(gemeinden)
frame.head()
```

```python
figure = build_choropleth(gemeinden, frame)
figure.show()
```
